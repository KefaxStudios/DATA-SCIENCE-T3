```python
# search for a value in the price column with 2 decimal places
df["price"] = df["price"].astype(str).str.replace(r"(?<=\d)\.(?=.*\.)", "", regex=True)
df["shipping"] = df["shipping"].astype(str).str.replace(r"[^\d.]", "", regex=True)
```

```python
# Function to clean the Price column
def price(price):
    price = str(price).replace(",", "")  # Remove commas
    if "to" in price:
        price = price.split("to")[0]  # Take the first value in a range
    try:
        return float(price)  # Convert to float
    except ValueError:
        return None  # Handle non-convertible values

# Apply cleaning function
df["price"] = df["price"].apply(price)
```

```python
def price(price):
    if pd.isna(price) or price == "":  # Handle missing or empty values
        return None  # Or use 0 if you prefer
    price = str(price).replace(",", "").strip()  # Remove commas and trim spaces
    if "to" in price:
        price = price.split("to")[0]  # Take the first value in a range
    try:
        return float(price)  # Convert to float
    except ValueError:
        return None  # Handle any remaining invalid values

# Apply cleaning function
df["price"] = df["price"].apply(price)
```

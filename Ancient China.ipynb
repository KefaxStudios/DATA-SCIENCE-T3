```python
import re
from calendar import month_name

# Create a dictionary mapping month names to numbers
month_to_num = {name.lower(): f"{i:02}" for i, name in enumerate(month_name) if i}

# Patterns for date formats
full_date_pattern = re.compile(
    r'\b(' + '|'.join(month_to_num.keys()) + r')\s+(\d{1,2})\s+(\d{4})\b', flags=re.IGNORECASE
)
month_year_pattern = re.compile(
    r'\b(' + '|'.join(month_to_num.keys()) + r')\s+(\d{4})\b', flags=re.IGNORECASE
)

# Sample text
sample_text = "The shuttle mission was completed on March 9 2011 and another mission happened in July 2011."

# Find matches
full_matches = full_date_pattern.findall(sample_text)
month_year_matches = month_year_pattern.findall(sample_text)

print("Full Date Matches:", full_matches)
print("Month-Year Matches:", month_year_matches)
```

```python
# Regex for YYYYMMDD dates
yyyymmdd_pattern = re.compile(r'\b(20\d{6})\b')

def standardize_full_date(match):
    """Convert 'March 9 2011' to '2011-03-09'."""
    month_str, day, year = match.groups()
    month_num = month_to_num.get(month_str.lower(), "00")  # Default to '00' if not found
    day_formatted = f"{int(day):02}"  # Ensure day is two digits
    return f"{year}-{month_num}-{day_formatted}"

def standardize_month_year(match):
    """Convert 'July 2011' to '2011-07'."""
    month_str, year = match.groups()
    month_num = month_to_num.get(month_str.lower(), "00")
    return f"{year}-{month_num}"


def standardize_yyyymmdd(match):
    """Convert 'YYYYMMDD' to 'YYYY-MM-DD'."""
    date_str = match.group(1)
    return f"{date_str[:4]}-{date_str[4:6]}-{date_str[6:]}"  # Format as YYYY-MM-DD

# Update process_dates function to include this new pattern
def process_dates(text):
    # Standardize full dates (e.g., "March 9 2011" -> "2011-03-09")
    text = full_date_pattern.sub(standardize_full_date, text)

    # Standardize month-year dates (e.g., "July 2011" -> "2011-07")
    text = month_year_pattern.sub(standardize_month_year, text)

    # Standardize YYYYMMDD format (e.g., "20220405" -> "2022-04-05")
    text = yyyymmdd_pattern.sub(standardize_yyyymmdd, text)

    return text
```

```python
df['content'] = df['content'].astype(str).apply(process_dates)
df['summary'] = df['summary'].astype(str).apply(process_dates)


```

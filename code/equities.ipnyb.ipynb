```python
df = pd.read_csv('equities_data.csv')
print(df.head())
print(df.tail())
```

```python
# dataframe information
print("\n--- DataFrame Info ---")
df.info()
```

```python
# checking for missing values
print(df.isna().sum())
```

```python
# creating category 'Missing' for missing values
for column in df.select_dtypes(include=['object', 'category']).columns:
    df[column] = df[column].fillna('Missing') 
```

```python
# checking for missing values after creating the 'Missing' category
print(df.isna().sum())
```

```python
# checking for duplicates
duplicate_rows = df[df.duplicated(keep = False)]
print("Duplicate Rows based on all columns:\n", duplicate_rows)

num_duplicate_rows = df.duplicated().sum()
print("Number of duplicate rows based on all columns:", num_duplicate_rows)
```

```python
#removing '\n'
df[column] = df[column].str.replace('\n', ' ')
print(df)
```

```python
#checking for \n after replacing.
num_with_newline = df['content'].str.contains('\n').sum()
num_with_newline = df['summary'].str.contains('\n').sum()
num_with_newline = df['links'].str.contains('\n').sum()
num_with_newline = df['title'].str.contains('\n').sum()
num_with_newline = df['url'].str.contains('\n').sum()
print(f"Number of rows with newline characters: {num_with_newline}")
```

```python
#finding words that have no spaces between them
def find_nospaces(text):
    pattern = r"([a-zA-Z0-9]+){2,}"  
    matches = re.findall(pattern, text)
    return matches
df['nospace_sequences_count'] = df['content'].apply(find_nospaces)
df['nospace_sequences_count'] = df['summary'].apply(find_nospaces)
df['nospace_sequences_count'] = df['title'].apply(find_nospaces)
total_no_space = df['nospace_sequences_count'].sum()
print(df)
print(f"Total words without space {total_no_space}")
```
